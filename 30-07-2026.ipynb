#### Program (6) - Implement stemming and Lemmatization

```python
import nltk
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

nltk.download('wordnet')

words = ["running", "jumps",  "happily",'unhappy', 'disagree','unwanted','mice']

porter_stemmer = PorterStemmer()

wordnet_lemmatizer = WordNetLemmatizer()

stemmed_words = [porter_stemmer.stem(word) for word in words]

lemmatized_words = [wordnet_lemmatizer.lemmatize(word,pos = 'v') for word in words]

print(f"{"Original word":<20}{"Stemmed word":<20}{"Lemmatized word":<20}")
print('---'*30)
for i in range(len(words)):
    print(f"{words[i]:<20}{stemmed_words[i]:<20}{lemmatized_words[i]:<20}")

```

#### Write a program to implement lemmatization with pos tagging using nltk.

```python
import nltk

from nltk.stem import WordNetLemmatizer

from nltk.corpus import wordnet

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')

nltk.download('wordnet')

def get_pos(tag):

    if tag.startswith('J'):
        return wordnet.ADJ

    elif tag.startswith('V'):
        return wordnet.VERB

    elif tag.startswith('N'):
        return wordnet.NOUN

    elif tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN

lemmatizer = WordNetLemmatizer()
sentence = "The children are running towards a better place."
tokens = word_tokenize(sentence)
pos_tags = nltk.pos_tag(tokens)

for word, tag in pos_tags:
    lemma = lemmatizer.lemmatize(word,get_pos(tag))
    print(f"{word:<20}{tag:<20} -> {lemma}")
    
```
