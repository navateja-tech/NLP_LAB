1.Given a text document solve the following tasks:  
    i. Count the number of words and the distinct words.  
    ii. Using list addition, and the set and sorted operations, compute the vocabulary of the sentences sent1 ... sent8.  
    iii. Define sent to be the list of words ['she', 'sells', 'sea', 'shells', 'by', 'the', 'sea', 'shore']. Now write code to perform the following tasks:  
    a. Print all words beginning with sh.   
    b. Print all words longer than four characters.  


```python
import os

words_cnt=0
words_set=set()
with open('example.txt','r') as file:
    lines=file.read().split('\n')
    for line in lines:
        words=line.split(' ')
        words_cnt+=len(words)
        for word in words:
            words_set.add(word)
    print(f"Number of words are {words_cnt} ")
    print(f"Number of distinct words are {len(words_set)}")
    
```

```python

words_set=set()
with open('example.txt','r') as file:
    lines=file.read().split('\n')
    for line in lines:
        words=line.split(' ')
        for word in words:
            words_set.add(word)
    vocabulary=sorted(list(words_set))
    
    print(f"Vocabulary is {vocabulary}")
```

```python
vocabulary=['she', 'sells', 'sea', 'shells', 'by', 'the', 'sea', 'shore']
words = [word for word in vocabulary if word.startswith('sh')]
print(f"Words starting with sh are {words}")

print(f"words longer than four characters are {[word for word in vocabulary if len(word)>4]}")


    
```

```python

text1="yeah! This is wonderful! Yes or not? what do think about this?"
text2="yeah! I think this is wonderful!"

s1=set(text1.split(' '))
s2=set(text2.split(' '))

print("Intersection - ",s1.intersection(s2))
print("Difference - ",s1-s2)
```

```python
from collections import Counter
text1="I am playing cricket in the evening, You can also play cricket with us"
d=Counter(text1.split(' '))
print("Dict : ",d)
cnt=0
word=''
for k,v in d.items():
    if v>cnt:
        word=k
        cnt=v
print(f"Most Occuring word is {word} with frquency= {cnt}")


```

```python
import re 
ph_pattern=r"[6-9][0-9]{9}"
email_pattern=r"\b[a-z][a-z0-9.]*\@gmail\.com\b"

info="I am Jack, I live in Australia, You can contact me. My details are phno- 7995289655 and email is willjacks01@gmail.com "
ph=re.search(ph_pattern,info)
email=re.search(email_pattern,info)

print(f"Phone number : {ph.group()  if ph is not None else None} and Email is {email.group() if email is not None else None}")
```
