### 10 - Find POS tags for the words in a sentence using viterbi decoding 

```python
def viterbi(words, tags, start_p, trans_p, emit_p):
    backpointer = [{}]

    # Step 1: Initialize first word probabilities
    for tag in tags:
        viterbi_table[0][tag] = start_p[tag] * emit_p[tag].get(words[0], 0.0)
        backpointer[0][tag] = None

    # Step 2: Fill Viterbi table for remaining words
    for t in range(1, len(words)):
        viterbi_table.append({})
        backpointer.append({})
        
        for curr_tag in tags:
            # Find max probability and previous tag state
            max_tr_prob, best_prev_tag = max(
                (viterbi_table[t-1][prev_tag] * trans_p[prev_tag].get(curr_tag, 0.0), prev_tag)
                for prev_tag in tags
            )
            
            # Multiply max transition path by emission probability
            viterbi_table[t][curr_tag] = max_tr_prob * emit_p[curr_tag].get(words[t], 0.0)
            backpointer[t][curr_tag] = best_prev_tag

    # Step 3: Find best path termination
    best_last_tag = max(viterbi_table[-1], key=viterbi_table[-1].get)
    
    # Step 4: Backtrack optimal path
    best_path = [best_last_tag]
    for t in range(len(words) - 1, 0, -1):
        best_last_tag = backpointer[t][best_last_tag]
        best_path.insert(0, best_last_tag)

    return list(zip(words, best_path))


# --- Data Setup ---
words = ["The", "dog", "barked"]
tags = ["D", "N", "V"]

start_p = {"D": 0.6, "N": 0.3, "V": 0.1}

trans_p = {
    "D": {"D": 0.0, "N": 0.5, "V": 0.1},
    "N": {"D": 0.0, "N": 0.2, "V": 0.6},
    "V": {"D": 0.1, "N": 0.3, "V": 0.0}
}

emit_p = {
    "D": {"The": 0.9, "dog": 0.0, "barked": 0.0},
    "N": {"The": 0.1, "dog": 0.8, "barked": 0.1},
    "V": {"The": 0.0, "dog": 0.1, "barked": 0.9}
}
viterbi_table = [{}]

# --- Run ---
result = viterbi(words, tags, start_p, trans_p, emit_p)
for item in viterbi_table:
    print(item)

print()
print("Predicted POS Tags:", result)
```
