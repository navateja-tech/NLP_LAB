## Program-2
#### Finding out the conditional frequency distributions for the selected genres and words from the Brown corpus

```python
!pip install nltk
```

```python
import nltk
from nltk.corpus import brown
from nltk.probability import ConditionalFreqDist

# nltk.download('brown')

genres=["news","romance","adventure","fiction"]
target_words=["the","love","government"]

cdf=ConditionalFreqDist(
    (genre,word.lower())
    for genre in genres
    for word in brown.words(categories=genre))
# print(cdf)

print(f"{'Genre/Words':<12}",end="")
for word in target_words:
    print(f"{word:<15}",end="")
print('\n',"-----"*20,end="\n")

for genre in genres:
    print(f"{genre:<12}",end="")
    for word in target_words:
        print(f"{cdf[genre][word]:<15}",end="")
    print("\n")
```

## Program-3

#### Write a Python program to demonstrate morphological analysis and morphological generation as inverse processes.

```python

def remove_suffix(word):
    if word.endswith('ing') or word.endswith('ful'):
        return word[:-3],word[-3:]
    elif word.endswith('ed') or word.endswith('er') or word.endswith('or'):
        return word[:-2],word[-2:]
    elif word.endswith('able') or word.endswith('ful'):
        return word[:-4],word[-4:]
    elif word.endswith('s'):
        return word[:-1],word[-1:]
    return word,'-'


def remove_prefix(word):
    if word.startswith('im') or word.startswith('un') or word.startswith('re'):
        return word[:2],word[2:]
    elif word.startswith('pre') or word.startswith('mis') or word.startswith('dis') :
        return word[:3],word[3:]
    return '-',word

words=['unplayable','sprinting','impossible','granted','unwanted','disbelief','waiter']
print(f"{'Words':<15}{'Prefix':<10}{'Root':<15}{'Suffix':<10}")
print("----"*20)
for word in words:
    prefix,root=remove_prefix(word)
    root,suffix=remove_suffix(root)
    print(f"{word:<15}{prefix:<10}{root:<15}{suffix:<10}",end="\n")
    

```
