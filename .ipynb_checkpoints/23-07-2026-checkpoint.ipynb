#### AIM 5 : Implement Tokenization and segmentation of the Text using Regular Expressions of NLTK

#### Tokenization is a foundational process of breaking down the raw text into smaller manageble words, it serves as a critical step in NLP because the machines cant understand the raw text ,so the text should be converted into tokens which are then converted into numerical identity numbers called embeddings 

```python
import requests
from bs4 import BeautifulSoup
from nltk.tokenize import RegexpTokenizer
import re

url = "https://www.thehindu.com/news/national/"

response = requests.get(url)

# print(response.text)

soup = BeautifulSoup(response.text,'html.parser')

# print(soup)

text = " ".join(p.get_text() for p in soup.find_all('p'))

# print(text)

# docs = text.split(' ')

# tokens=[]
# for t in docs:
#     if t:
#         val = re.sub('[?\n\t\.\xa0]','',t)
#         tokens.append(val)

tokenizer = RegexpTokenizer(r'\w+')

tokens = tokenizer.tokenize(text)


print(tokens)

```

#### Additional Program - Write a program to parse sentence using context free grammer

```python
from nltk.parse import ChartParser
from nltk import CFG

grammer = CFG.fromstring(""" 
    S -> NP VP
    NP -> Det N
    VP -> V NP
    Det -> 'the'
    V -> 'kicked'
    N -> 'boy' | 'ball'
""")

parser = ChartParser(grammer)

sentence = "the boy kicked the ball".split()

for tree in parser.parse(sentence):
    print(tree)
    # tree.pretty_print()
    tree.draw()
```
