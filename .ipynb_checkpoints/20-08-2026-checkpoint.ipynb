### Aim - 9 : Calculate emission and transition matrix

```python
import nltk
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger_eng')
```

```python

import nltk

sentences=["The dog is running","cat is drinking milk"," Iam tired"]
words = nltk.word_tokenize(sentence)


words_tags = nltk.pos_tag(words)

tags = set()

print("\nPOS Tags:")
for word, tag in words_tags:
    print(word, "->", tag)
    tags.add(tag)
    


for sentence in sentences:
    for word, tag in sentence:
        tags.add(tag)

tags = list(tags)

print("POS Tags:", tags)

# --------------------------------------------------
# 1. Calculate Transition Counts
# --------------------------------------------------

transition_count = defaultdict(lambda: defaultdict(int))

for sentence in sentences:
    for i in range(len(sentence) - 1):
        current_tag = sentence[i][1]
        next_tag = sentence[i + 1][1]

        transition_count[current_tag][next_tag] += 1

# --------------------------------------------------
# 2. Calculate Transition Probabilities
# --------------------------------------------------

print("\nTransition Matrix:")

for current_tag in tags:
    total = sum(transition_count[current_tag].values())

    for next_tag in tags:
        if total > 0:
            probability = transition_count[current_tag][next_tag] / total
        else:
            probability = 0

        print(
            f"P({next_tag} | {current_tag}) = {probability:.2f}"
        )

# --------------------------------------------------
# 3. Calculate Emission Counts
# --------------------------------------------------

emission_count = defaultdict(lambda: defaultdict(int))

for sentence in sentences:
    for word, tag in sentence:
        emission_count[tag][word] += 1

# --------------------------------------------------
# 4. Calculate Emission Probabilities
# --------------------------------------------------

print("\nEmission Matrix:")

for tag in tags:
    total = sum(emission_count[tag].values())

    for word in emission_count[tag]:
        probability = emission_count[tag][word] / total

        print(
            f"P({word} | {tag}) = {probability:.2f}"
        )
```
