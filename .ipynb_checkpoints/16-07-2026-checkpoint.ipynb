#### Practice Programs

```python
import nltk
nltk.download('punkt')

from nltk.tokenize import word_tokenize,sent_tokenize
text = """Natural Language Processing is a branch of AI.
NLTK provides tools for tokenization, stemming, and parsing.
"""
sent_tokens = sent_tokenize(text)
word_tokens = word_tokenize(text)

for sent in sent_tokens:
    print(sent)
    
print()

for word in word_tokens:
    print(word)
```

#### Write a program to caluculate word frequency.

```python

import nltk
from nltk import FreqDist
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')

text = """
NLP is a branch of AI.
NLP
helps computers understand human language.
"""

words = word_tokenize(text)

words = [word.lower() for word in words if word.isalpha()]

freq_dist = FreqDist(words)

print("Words:")
print(words)

print("\n" + "-" * 50)

print("Word Frequency:")
for word, frequency in freq_dist.items():
    print(f"{word} : {frequency}")
```
